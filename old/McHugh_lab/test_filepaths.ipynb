```python
import os
from collections import defaultdict

directory = '/groups/cns/share/Vincent/McHugh_lab/CA2Cre_DREADD/extracted_spike_trains/'
files = os.listdir(directory)
print(files)

# Parse filenames and group by (name, day)
groups = defaultdict(list)
for file in files:
    if file.endswith('.npy'):
        parts = file.split('_')
        if len(parts) >= 6:
            name, day, task, tetrode, area, treatment = parts[0], parts[1], parts[2], parts[3], parts[4], parts[5].replace('.npy', '')
            groups[(name, day)].append((task, treatment))

# Find animal-day combinations where task=lt1 exists for both pre and post
selected_combinations = []
for (name, day), entries in groups.items():
    has_pre = any(task == 'lt1' and treatment == 'pre' for task, treatment in entries)
    has_post = any(task == 'lt1' and treatment == 'post' for task, treatment in entries)
    if has_pre and has_post:
        selected_combinations.append((name, day))

print("Selected animal-day combinations:")
for combo in selected_combinations:
    print(combo)
```
